ANALYZYING THIS HEALTHCARE DATASET

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Data importing

```python
df = pd.read_csv("healthcare_dataset_.csv")
df.head()
```

Data checking

```python
df.shape
```

```python

df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
```

```python
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])
```

```python
df['billing_amount'] = pd.to_numeric(df['billing_amount'], errors='coerce')
```

```python
df.info()
```

```python
df.describe()
```

Data Cleaning

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
df.isnull().sum()
```

Data formatting

```python
df['name'] = df['name'].str.title()
```

```python
df['date_of_admission'] = pd.to_datetime(df['date_of_admission'])
```

```python
df['discharge_date'] = pd.to_datetime(df['discharge_date'])
```

```python
df['length_of_stay'] = (df['discharge_date'] - df['date_of_admission']).dt.days
```

```python
df.head()
```

Age check analysis

```python
df['age_group'] = pd.cut(df['age'], 
                        bins=[0,18,35,60,100],
                        labels=['Child','Young Adult','Adult','Elderly'])
```

```python
df[df['age_group'].isnull()][['name', 'age', 'age_group']]
```

```python
print(df['age_group'].value_counts())
```

Columns for day and months

```python
df['admission_month'] = df['date_of_admission'].dt.strftime('%B')
```

```python
df['day_of_week'] = df['date_of_admission'].dt.strftime('%A')
```
